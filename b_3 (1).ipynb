```python
##key features(according to my domain knowldge)
##length_url=Phishing URLs are usually longer to hide malicious parameters.
##nb_dots = More dots (.) in a URL often suggest a subdomain trick.
##nb_hyphens = Attackers use hyphens to mimic legitimate websites.
##nb_www= Extra www in a URL could indicate a deceptive structure (e.g., www.secure-login.www.bank.com).
##ratio_digits_url = Phishing URLs often contain more numbers to look legitimate (secure-bank123.com or paypal-login100.com).
```

```python
key_features = ['length_url', 'nb_dots', 'nb_hyphens', 'nb_www', 'ratio_digits_url']
df[key_features].hist(figsize=(12,8), bins=30, color='skyblue')
plt.suptitle("Feature Distributions", fontsize=16)
plt.show()
##length url = Most URLs are short, with a sharp decline in frequency as the length increases.
##nb_dots = Most URLs have very few dots (0–5). hishing URLs might use multiple dots to disguise their identity
##nb_hyphens =  high count of hyphens in some URLs might indicate potential phishing attempts.
###The distribution suggests that many URLs use a non-standard format.
## ratio_digits = Most URLs have low numeric content, but some have a relatively high proportion of digits.
```

```python
sns.pairplot(df, hue='status', vars=key_features, palette='coolwarm')
plt.show()
##length_url = Phishing URLs (orange) tend to be longer than legitimate ones.
##The distribution is highly right-skewed, meaning some phishing URLs are extremely long.

##nb_dots & nb_hyphens = Phishing URLs have a higher number of dots and hyphens compared to legitimate ones.
##More dots and hyphens suggest an attempt to mimic legitimate domains (secure-login.bank.com vs. bank.com).
##A clear separation between phishing and legitimate URLs is visible in scatter plots.

###nb_www = This feature is mostly binary (0 or 1).Phishing URLs may exclude "www" to disguise themselves.

##ratio_digits_url = Phishing URLs tend to have a higher ratio of digits compared to legitimate ones.
##Some phishing links use random numbers to appear legitimate (e.g., securebank123-login.com).
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df["status"]=le.fit_transform(df["status"])
##0= legitimate
##1= phishing
```

```python
df["status"]
```

```python
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
df.iloc[:,1:-1]=ss.fit_transform(df.iloc[:,1:-1])
```

```python
df.iloc[:,1:-1]
```

```python
features=df.drop(columns=["url","status"])
target=df["status"]
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(features,target,random_state=42,test_size=0.2,stratify=target)
print(f"Training data : {xtrain.shape, ytrain.shape}")
print(f"Testing data : {xtest.shape, ytest.shape}")
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(xtrain,ytrain)
y_pred=lr.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
ac= accuracy_score(ytest,y_pred)
cr=classification_report(ytest,y_pred)
cm=confusion_matrix(ytest,y_pred)
print(f"accuracy score : {ac}\nconfusion matrix :\n{cm}\nclassification report :\n{cr}")
```

```python
print(f"Training score : {lr.score(xtrain,ytrain)}")
print(f"Testing score : {lr.score(xtest,ytest)}")
```

```python
from sklearn.metrics import roc_auc_score,roc_curve
ra=roc_auc_score(ytest,y_pred)
fpr,tpr,thres=roc_curve(ytest,y_pred)
plt.plot(fpr,tpr)
plt.title("ROC curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.grid()
plt.show()
ra
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
dt.fit(xtrain,ytrain)
y_pred_dt=dt.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
ac= accuracy_score(ytest,y_pred_dt)
cr=classification_report(ytest,y_pred_dt)
cm=confusion_matrix(ytest,y_pred_dt)
print(f"accuracy score : {ac}\nconfusion matrix :\n{cm}\nclassification report :\n{cr}")
```

```python
from sklearn.metrics import precision_recall_curve
```

```python
# Compute ROC curve
fpr, tpr, _ = roc_curve(ytest, y_proba)

# Compute Precision-Recall curve
precision, recall, _ = precision_recall_curve(ytest, y_proba)

# Plot both curves
plt.figure(figsize=(6,4))

# ROC Curve
plt.plot(fpr, tpr, label="ROC Curve (AUC)", color="blue")
# Plot Precision-Recall Curve
plt.plot(recall, precision, label="Precision-Recall Curve", color="red")
# Diagonal line for ROC curve reference
plt.plot([0, 1], [0, 1], linestyle="--", color="gray")

# Labels & Title
plt.xlabel("False Positive Rate (ROC) / Recall (PR)")
plt.ylabel("True Positive Rate (ROC) / Precision (PR)")
plt.title("ROC Curve & Precision-Recall Curve")
plt.legend()
plt.show()
```
